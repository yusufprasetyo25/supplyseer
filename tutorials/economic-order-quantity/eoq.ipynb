# Economic Order Quantity (EOQ)

This tutorial implements a EOQ. The model uses traditional economic order quantity with its parameters (demand, order cost, holding cost).

Assumptions [(Hamdy A. Taha, 1987)](https://dl.acm.org/doi/book/10.5555/27156):
- Demand is constant over time with no shortages. 
- Total cost per unit time = setup cost per unit time + holding cost per unit time

Key Features:
- Generates the economic order quantity from the parameters

The Economic Order Quantity provides:
- Economic order quantity values from the parameters

Library Dependencies:
- numpy: Numerical computations

### Scenario

The scenario is subset of example from [Hamdy A. Taha: Operations Research, 4th Edition](https://dl.acm.org/doi/book/10.5555/27156), page 511:

**Example 13.3-1.** The daily demand for a commodity is approximately 100 units. Every time an order is placed, a fixed cost of \$100 is incurred. The daily holding cost per unit inventory is \$.02. Determine the economic lot size.

```python
from supplyseer.eoq import eoq
```

We are going to calculate the EOQ and the cycle length for better understanding of the example

```python
eoq(d=100, a=100, h=0.02)
```

The economic order quantity is 1000 units per order, while the daily demand is 100 units. Thus, the cycle length for every order is 10 days.

From this particular example, we get the general framework of how we can use the EOQ:
1. We calculate the optimum order quantity from the given parameters from EOQ formula
2. We operationalize the result through calculating the cycle length from demand
